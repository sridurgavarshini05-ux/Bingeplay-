```python

query_q3 = """
SELECT
    device_type,
    COUNT(*) AS total_sessions,
    SUM(watch_minutes) AS total_watch_minutes,
    ROUND(AVG(watch_minutes), 2) AS avg_watch_minutes,
    ROUND(
        100.0 * SUM(CASE WHEN completed = 1 THEN 1 ELSE 0 END)
        / COUNT(*), 2
    ) AS completion_rate_pct
FROM watch_sessions
WHERE user_id IS NOT NULL
GROUP BY device_type
ORDER BY total_watch_minutes DESC;
"""

df_q3 = pd.read_sql(query_q3, engine)
print(df_q3.to_string(index=False))
```

```python

query_q6 = """
WITH binge_days AS (
    SELECT
        user_id,
        show_id,
        session_date,
        COUNT(*) AS session_count
    FROM watch_sessions
    WHERE user_id IS NOT NULL
      AND session_date >= '2024-04-01'
      AND session_date < '2024-07-01'
    GROUP BY user_id, show_id, session_date
    HAVING COUNT(*) >= 5
),
user_binge_counts AS (
    SELECT
        user_id,
        COUNT(*) AS binge_day_count
    FROM binge_days
    GROUP BY user_id
)
SELECT
    (SELECT COUNT(*) FROM binge_days) AS total_binge_days,
    user_id AS top_user_id,
    binge_day_count AS top_user_binge_days
FROM user_binge_counts
ORDER BY binge_day_count DESC, user_id
LIMIT 1;
"""

df_q6 = pd.read_sql(query_q6, engine)
print(df_q6.to_string(index=False))
```

## Q7 — Q1 Signups Who Never Watched

**Total Q1 signups:** 1,250

**Q1 signups who never watched anything:** 226

**Interpretation:** Out of 1,250 users who signed up between January and March 2024, 226 had no recorded watch sessions. This represents 18.08% of Q1 signups and may indicate an opportunity to improve user onboarding and encourage first-time viewing.

**SQL approach:** Used `NOT EXISTS` to correctly identify users without any matching watch sessions, avoiding the `NULL` trap associated with `NOT IN`.

```python

query_q8 = """
WITH ranked_subscriptions AS (
    SELECT
        user_id,
        Plan,
        start_date,
        end_date,
        status,
        ROW_NUMBER() OVER (
            PARTITION BY user_id
            ORDER BY start_date DESC, subscription_id DESC
        ) AS rn
    FROM subscriptions
    WHERE start_date <= '2024-06-30'
      AND status = 'active'
      AND (end_date IS NULL OR end_date > '2024-06-30')
)
SELECT COUNT(*) AS overpaying_users
FROM ranked_subscriptions rs
WHERE rs.rn = 1
  AND rs.Plan IN ('Premium', 'Family')
  AND EXISTS (
      SELECT 1
      FROM watch_sessions ws
      WHERE ws.user_id = rs.user_id
  )
  AND NOT EXISTS (
      SELECT 1
      FROM watch_sessions ws
      JOIN shows s ON s.show_id = ws.show_id
      WHERE ws.user_id = rs.user_id
        AND s.min_plan IN ('Premium', 'Family')
  );
"""

df_q8 = pd.read_sql(query_q8, engine)
print(df_q8.to_string(index=False))
```

## Q8 — The Over-paying Premium/Family Users

**Answer:** 6 users.

**Interpretation:** Six users with a current Premium or Family subscription as of June 30, 2024, have recorded watch histories containing only Basic-tier shows. These users may be suitable candidates for targeted downgrade offers, which could help them choose a subscription plan that better matches their viewing habits.

**SQL approach:** Identified the latest eligible active subscription for each user and used `NOT EXISTS` to exclude users who watched any Premium- or Family-tier show.

```python

query_q9 = """
WITH ranked_subscriptions AS (
    SELECT
        s.*,
        ROW_NUMBER() OVER (
            PARTITION BY user_id
            ORDER BY start_date, subscription_id
        ) AS rn
    FROM subscriptions s
),
first_upgrades AS (
    SELECT
        user_id,
        MIN(start_date) AS first_upgrade_date
    FROM subscriptions
    WHERE Plan IN ('Premium', 'Family')
    GROUP BY user_id
),
active_users AS (
    SELECT DISTINCT user_id
    FROM subscriptions
    WHERE status = 'active'
      AND start_date <= '2024-06-30'
      AND (end_date IS NULL OR end_date > '2024-06-30')
)
SELECT
    COUNT(*) AS qualifying_users,
    ROUND(
        AVG(DATEDIFF(fu.first_upgrade_date, u.signup_date)),
        2
    ) AS avg_days_to_first_upgrade
FROM users u
JOIN ranked_subscriptions rs
    ON rs.user_id = u.user_id
   AND rs.rn = 1
JOIN first_upgrades fu
    ON fu.user_id = u.user_id
JOIN active_users au
    ON au.user_id = u.user_id
WHERE u.signup_date >= '2024-01-01'
  AND u.signup_date < '2024-02-01'
  AND rs.Plan = 'Basic'
  AND fu.first_upgrade_date > rs.start_date;
"""

df_q9 = pd.read_sql(query_q9, engine)
print(df_q9.to_string(index=False))
```

```python

query_q9 = """
WITH ranked_subscriptions AS (
    SELECT
        s.*,
        ROW_NUMBER() OVER (
            PARTITION BY user_id
            ORDER BY start_date, subscription_id
        ) AS rn
    FROM subscriptions s
    WHERE start_date <= '2024-06-30'
),
first_subscriptions AS (
    SELECT user_id, Plan, start_date
    FROM ranked_subscriptions
    WHERE rn = 1
),
first_upgrades AS (
    SELECT
        fs.user_id,
        MIN(s.start_date) AS first_upgrade_date
    FROM first_subscriptions fs
    JOIN subscriptions s
      ON s.user_id = fs.user_id
     AND s.start_date > fs.start_date
     AND s.Plan IN ('Premium', 'Family')
    WHERE fs.Plan = 'Basic'
    GROUP BY fs.user_id
),
active_users AS (
    SELECT DISTINCT user_id
    FROM subscriptions
    WHERE status = 'active'
      AND start_date <= '2024-06-30'
      AND (end_date IS NULL OR end_date > '2024-06-30')
)
SELECT
    COUNT(*) AS qualifying_users,
    ROUND(
        AVG(DATEDIFF(fu.first_upgrade_date, u.signup_date)), 2
    ) AS avg_days_to_first_upgrade
FROM users u
JOIN first_subscriptions fs
  ON fs.user_id = u.user_id
JOIN first_upgrades fu
  ON fu.user_id = u.user_id
JOIN active_users au
  ON au.user_id = u.user_id
WHERE u.signup_date >= '2024-01-01'
  AND u.signup_date < '2024-02-01'
  AND fs.Plan = 'Basic';
"""

df_q9 = pd.read_sql(query_q9, engine)
print(df_q9.to_string(index=False))
```

## Q9 — Upgrade Success Cohort

**Qualifying users:** 55

**Average days from signup to first upgrade:** 64.96 days

**Interpretation:** The analysis identifies 55 users who signed up in January 2024, started with a Basic subscription, later upgraded to Premium or Family, and had an active subscription as of June 30, 2024. On average, these users took approximately 65 days from signup to their first higher-tier subscription.

```python

query_q10 = """
WITH comeback_events AS (
    SELECT DISTINCT
        ws1.user_id,
        ws1.show_id,
        ws1.session_date AS incomplete_date
    FROM watch_sessions ws1
    WHERE ws1.user_id IS NOT NULL
      AND ws1.completed = 0
      AND EXISTS (
          SELECT 1
          FROM watch_sessions ws2
          WHERE ws2.user_id = ws1.user_id
            AND ws2.show_id = ws1.show_id
            AND ws2.session_date BETWEEN
                DATE_ADD(ws1.session_date, INTERVAL 1 DAY)
                AND DATE_ADD(ws1.session_date, INTERVAL 7 DAY)
      )
),
show_comebacks AS (
    SELECT
        ce.show_id,
        COUNT(*) AS comeback_count
    FROM comeback_events ce
    GROUP BY ce.show_id
),
ranked_shows AS (
    SELECT
        sc.*,
        ROW_NUMBER() OVER (
            ORDER BY comeback_count DESC, show_id
        ) AS rn
    FROM show_comebacks sc
)
SELECT
    (SELECT COUNT(*) FROM comeback_events) AS total_comeback_events,
    rs.show_id AS top_show_id,
    sh.title AS top_show_title,
    rs.comeback_count AS top_show_comebacks
FROM ranked_shows rs
JOIN shows sh ON sh.show_id = rs.show_id
WHERE rs.rn = 1;
"""

df_q10 = pd.read_sql(query_q10, engine)
print(df_q10.to_string(index=False))
```

### Q10 — Cliffhanger Comebacks: Results and Interpretation

**Results:**
- Total cliffhanger comeback events: **4,345**
- Show with the most comeback events: **Rayalaseema Raga** (`S088`)
- Comeback events for this show: **64**

**Interpretation:**

The analysis identified 4,345 unique cliffhanger comeback events. These events occur when a user has an incomplete viewing session for a show and watches the same show again within the following 1–7 days. *Rayalaseema Raga* had the highest number of comeback events, with 64, indicating repeated returns to the show after incomplete sessions.

```python

query_q11 = """
WITH user_weeks AS (
    SELECT DISTINCT
        user_id,
        DATE_SUB(session_date, INTERVAL WEEKDAY(session_date) DAY) AS week_start
    FROM watch_sessions
    WHERE user_id IS NOT NULL
),
numbered AS (
    SELECT
        user_id,
        week_start,
        ROW_NUMBER() OVER (
            PARTITION BY user_id
            ORDER BY week_start
        ) AS rn,
        DATEDIFF(week_start, '2000-01-03') DIV 7 AS week_ord
    FROM user_weeks
),
islands AS (
    SELECT
        user_id,
        week_start,
        week_ord - rn AS island_id
    FROM numbered
),
streaks AS (
    SELECT
        user_id,
        island_id,
        COUNT(*) AS streak_weeks
    FROM islands
    GROUP BY user_id, island_id
),
user_streaks AS (
    SELECT
        user_id,
        MAX(streak_weeks) AS longest_streak
    FROM streaks
    GROUP BY user_id
)
SELECT
    (
        SELECT COUNT(*)
        FROM user_streaks
        WHERE longest_streak >= 4
    ) AS users_4plus_streak,
    user_id AS longest_streak_user,
    longest_streak
FROM user_streaks
WHERE longest_streak = (
    SELECT MAX(longest_streak)
    FROM user_streaks
)
ORDER BY user_id
LIMIT 1;
"""

df_q11 = pd.read_sql(query_q11, engine)
print(df_q11.to_string(index=False))
```

### Q11 — Consecutive Weekly Viewing Streaks: Results and Interpretation

**Results:**
- Users with at least 4 consecutive active weeks: **1,675**
- User with the longest streak: **U00213**
- Longest streak: **26 consecutive weeks**

**Interpretation:**

The analysis identified 1,675 users who watched at least one session during four or more consecutive calendar weeks. User `U00213` had the longest viewing streak, with activity across 26 consecutive weeks. This indicates sustained engagement with the BingePlay platform over an extended period.

```python

query_q12 = """
WITH monthly_watch AS (
    SELECT
        user_id,
        SUM(
            CASE
                WHEN session_date >= '2024-05-01'
                 AND session_date < '2024-06-01'
                THEN watch_minutes
                ELSE 0
            END
        ) AS may_minutes,
        SUM(
            CASE
                WHEN session_date >= '2024-06-01'
                 AND session_date < '2024-07-01'
                THEN watch_minutes
                ELSE 0
            END
        ) AS june_minutes
    FROM watch_sessions
    WHERE user_id IS NOT NULL
      AND session_date >= '2024-05-01'
      AND session_date < '2024-07-01'
    GROUP BY user_id
),
churn_signals AS (
    SELECT
        user_id,
        may_minutes,
        june_minutes,
        ROUND(
            100.0 * (may_minutes - june_minutes) / may_minutes,
            2
        ) AS drop_percentage
    FROM monthly_watch
    WHERE may_minutes > 0
      AND june_minutes <= may_minutes * 0.50
)
SELECT
    cs.user_id,
    u.Name AS user_name,
    cs.may_minutes,
    cs.june_minutes,
    cs.drop_percentage,
    (SELECT COUNT(*) FROM churn_signals) AS total_users_with_drop
FROM churn_signals cs
JOIN users u
    ON u.user_id = cs.user_id
ORDER BY cs.drop_percentage DESC, cs.user_id;
"""

df_q12 = pd.read_sql(query_q12, engine)
print(df_q12.to_string(index=False))
```
